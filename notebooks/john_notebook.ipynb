## Load Packages

```python
import os
from dotenv import load_dotenv
import anthropic
import re
import pandas as pd
import numpy as np
```

# Import API Key

```python
load_dotenv()
api_key = os.getenv("API_KEY")
```

# Test Connection

```python
client = anthropic.Anthropic(
    # defaults to os.environ.get("ANTHROPIC_API_KEY")
    api_key=api_key,
)
message = client.messages.create(
    model="claude-3-5-sonnet-20240620",
    max_tokens=1024,
    messages=[
        {"role": "user", "content": "Hello, Claude"}
    ]
)
print(message.content)
```

# Create Model Class

```python
class claude():
    def __init__(self, api_key = api_key, model = "claude-3-5-sonnet-20240620"):
        self.client = anthropic.Anthropic(api_key = api_key)
        self.model = model

    def generate(self, prompt, max_tokens = 2048):
        response = self.client.messages.create(
            model=self.model,
            max_tokens=max_tokens,
            messages=[
                {"role": "user", "content": prompt}
            ]
        )
        return response
```

```python
claude_example = claude()
print(claude_example.generate("Hello, Claude"))
```

```python
claude_model = claude_example
```

# Create Test Questions/Data

```python
prompt = "Generate 100 random questions, and answers. These can be simple facts, or short response based answers."
```

```python
questions = claude_model.generate(prompt)
```

```python
print(questions)
```

```python
text = questions.content[0].text
```

```python
pattern = r'\d+\.\s+Q:\s*(.*?)\n\s+A:\s*(.*)'

matches = re.findall(pattern, text)

df = pd.DataFrame(matches, columns=["Question", "Answer"])
```
