This notebook demonstrates the process of setting up and running evaluations using the `promptfoo` framework through the terminal (not through this notebook).
Note: If you don’t want to install promptfoo globally, you can use `npx` to run it without installation which is what I did in this case.

To start, I will evaluate two key metrics:

1. Accuracy of Response
2. Bias Detection
3. Toxicity Detection

For more details, visit the [Promptfoo Getting Started Guide](https://www.promptfoo.dev/docs/getting-started/).

# Setup API Keys

This command reads the `.env` file from the directory (../../) where I have my `.env`, extracts the variables, and exports them to the environment:

`export $(cat ../../.env | xargs)`

## Initialize Promptfoo

This will create a `promptfooconfig.yaml` file in your current directory:

`npx promptfoo@latest init`

## Run Evaluations
This command will execute the evaluation based on the configuration in your `.yaml` file:

`npx promptfoo@latest eval`

I created different yaml config files for each metric I'm exploring

### To run evaluation for accuracy of response
`npx promptfoo@latest eval --config configs/accuracy_eval.yaml`

### To run evaluation for bias detection
`npx promptfoo@latest eval --config configs/bias_detection.yaml`

### To run evaluation for toxicity detection
`npx promptfoo@latest eval --config configs/toxicity_detection_binary.yaml`

## Share Results

We can share results via promptfoo's sharing feature, use: 

`npx promptfoo@latest share`
