Completions vs Messages:
- https://docs.anthropic.com/en/api/migrating-from-text-completions-to-messages
- With Messages, the response is the content value, which is a list of content blocks
- With Text Completions, you can pre-fill part of Claude’s response:
    - With Messages, you can achieve the same result by making the last input message have the assistant role:

````python
load_dotenv()
api_key = os.getenv("ANTHROPIC_API_KEY")

class LLM():
    def __init__(self, api_key = api_key, model = "claude-3-7-sonnet-20250219"):
        self.client = anthropic.Anthropic(api_key = api_key)
        self.model = model

    def generate(self, prompt, max_tokens = 1024):
        response = self.client.messages.create(
            model=self.model,
            max_tokens=max_tokens,
            messages=[
                {"role": "user", "content": prompt}
            ]
        )
        return response.content[0].text
````

````python
claude_example = LLM()
print(claude_example.generate("What model Claude are you?"))
````
