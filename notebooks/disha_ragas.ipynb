```python
client = anthropic.Anthropic(api_key=api_key)
```

```python
data = {
    "user_input": [
        "What is the name of the main female character in 'New Girl'?",
        "Which character in 'New Girl' is known for his eccentric behavior and cat, Ferguson?",
        "What is the profession of Jess in the early seasons of 'New Girl'?",
        "What game do the roommates frequently play that involves elaborate rules?"
    ],
    "retrieved_contexts": [
        ["'New Girl' follows the story of a quirky woman who moves in with three men after a breakup."],
        ["One of the 'New Girl' characters has a cat named Ferguson and is known for his over-the-top personality."],
        ["In the early seasons, Jess works at a job that aligns with her upbeat and nurturing personality."],
        ["The roommates in 'New Girl' often play a game with complex rules, which becomes a recurring theme in the show."]
    ],
    "reference": [
        "The main female character in 'New Girl' is Jess Day.",
        "The character known for his eccentric behavior and cat, Ferguson, is Winston Bishop.",
        "Jess works as a school teacher in the early seasons of 'New Girl'.",
        "The game the roommates frequently play is True American."
    ],
    "response": [
        "The main female character is Jess Day.",                # Correct
        "The character is Nick Miller.",                        # Incorrect
        "Jess works as a lawyer.",                              # Incorrect
        "The game is True American."                            # Correct
    ]
}

dataset = Dataset.from_dict(data)
```

```python
llm_client = ChatAnthropic(
    model="claude-2",
    anthropic_api_key=api_key
)

llm = LangchainLLMWrapper(llm_client)

metrics = [
    LLMContextRecall(llm=llm),
    FactualCorrectness(llm=llm),
    Faithfulness(llm=llm)
]

results = evaluate(
    dataset=dataset,
    metrics=metrics
)

for metric_name, score in results.items():
    print(f"{metric_name}: {score:.4f}")
```
