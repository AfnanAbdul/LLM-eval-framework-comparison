```python
from rouge_score import rouge_scorer
import nltk
from nltk.translate.bleu_score import sentence_bleu
from nltk.translate.meteor_score import meteor_score

# Ensure that you have downloaded necessary NLTK data
nltk.download('punkt')

# Sample data
generated_text = "The quick brown fox jumps over the lazy dog."
reference_text = "A fast brown fox leaps over a sleepy dog."

# 1. ROUGE Score
scorer = rouge_scorer.RougeScorer(["rouge1", "rouge2", "rougeL"], use_stemmer=True)
rouge_scores = scorer.score(reference_text, generated_text)
print("ROUGE Scores:", rouge_scores)

# 2. BLEU Score
reference_tokens = [reference_text.split()]
generated_tokens = generated_text.split()
bleu_score = sentence_bleu(reference_tokens, generated_tokens)
print("BLEU Score:", bleu_score)

# 3. METEOR Score
meteor_score_result = meteor_score([reference_text], generated_text)
print("METEOR Score:", meteor_score_result)
```
