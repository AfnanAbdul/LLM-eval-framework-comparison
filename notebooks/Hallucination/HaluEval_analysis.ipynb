```python
import pandas as pd
import copy
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
# Run command in terminal
# python HaluEval_evaluation.py --task qa --model claude-3-5-haiku-20241022
```

```python
file_path = "../../results/Hallucination/HaluEval/qa/qa_claude-3-5-haiku-20241022_results_sample.json"
qa_results_df = pd.read_json(file_path, lines=True)
```

```python
qa_results_df
```

```python
qa_results = copy.deepcopy(qa_results_df)

# Convert 'Yes' to 1 and 'No' to 0
qa_results["ground_truth"] = qa_results["ground_truth"].map({"Yes": 1, "No": 0})
qa_results["judgement"] = qa_results["judgement"].map({"Yes": 1, "No": 0})

# Compute metrics
accuracy = accuracy_score(qa_results["ground_truth"], qa_results["judgement"])
precision = precision_score(qa_results["ground_truth"], qa_results["judgement"])
recall = recall_score(qa_results["ground_truth"], qa_results["judgement"])
f1 = f1_score(qa_results["ground_truth"], qa_results["judgement"])

# Print results
print(f"Accuracy: {accuracy*100:.2f}%")
print(f"Precision: {precision*100:.2f}%")
print(f"Recall: {recall*100:.2f}%")
print(f"F1 Score: {f1*100:.2f}%")
```
