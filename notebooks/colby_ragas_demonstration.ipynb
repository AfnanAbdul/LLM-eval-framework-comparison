```python
import anthropic
import os
from dotenv import load_dotenv
from ragas.metrics import Faithfulness, AnswerRelevancy, ContextRelevancy, ContextRecall
from ragas.langchain.evalchain import RagasEvaluatorChain
from langchain_community.chat_models import ChatAnthropic
from ragas.metrics import (
    faithfulness,
    answer_relevancy,
    context_relevancy,
    context_recall,
)
```

```python
load_dotenv()
api_key = os.getenv("ANTHROPIC_API_KEY")
```

```python
claude = ChatAnthropic(model = 'claude-2', anthropic_api_key=api_key)
```

Knowledge bases are contained in text documents, and they contain information about a specific domain, to help a contextualized chat bot. RAG is method to preprocess the knowledge base and make it more efficient to search through, chunking the text into smaller pieces, turning them into semantic representations, and indexing them. Retrievers are used to search through the indexed knowledge base and find the most relevant chunks of text to answer a question.

Your application quality is a function both of the LLM you choose and the prompting and data retrieval strategies you employ to provide model context.

```python
#don't understand why it complains about openai key when we provide the llm as claude

eval_chains = {
    metric.name: RagasEvaluatorChain(metric=metric, llm = claude)
    for metric in [faithfulness, answer_relevancy, context_relevancy, context_recall]
}
```
